```python
vcf_file = cyvcf2.VCF("/home/hain/EMBL/Saturn3/data/sv_calls/test/merge.vcf")
filt_vcf_file = cyvcf2.Writer("/home/hain/EMBL/Saturn3/data/sv_calls/test/merge.SUPP2.vcf", vcf_file)

for record in vcf_file:
    
    support_manta = int(record.INFO["SUPP_VEC_EXT"][0]) + int(record.INFO["SUPP_VEC_EXT"][1])
    if support_manta > 1:
        support_manta = 1
    support_tiddit = int(record.INFO["SUPP_VEC_EXT"][2])

    if (support_manta + support_tiddit) >= 2:
        filt_vcf_file.write_record(record)
        
filt_vcf_file.close()
```
